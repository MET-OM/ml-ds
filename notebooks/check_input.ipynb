```python
import glob
from pathlib import Path

import pandas as pd
import xarray as xr
```

```python
def get_stats(da):
    mean_vals = da.mean(dim=["latitude", "longitude"])
    max_vals  = da.max(dim=["latitude", "longitude"])
    min_vals  = da.min(dim=["latitude", "longitude"])
    std_vals  = da.std(dim=["latitude", "longitude"])

    return pd.DataFrame({
        "mean": mean_vals.values,
        "max":  max_vals.values,
        "min":  min_vals.values,
        "std":  std_vals.values,
    }, index=mean_vals["valid_time"].values)
```

```python
DIR_DATA = str(Path.home() / "ml-ds_data")
ds_x = xr.open_mfdataset(f"{DIR_DATA}/*_reinterp.nc", combine="by_coords")
files = [f for f in glob.glob(f"{DIR_DATA}/ERA5*.nc") if "_reinterp" not in f]
ds_y = xr.open_mfdataset(files, combine="by_coords")
ds_gebco = xr.open_dataset(f"{DIR_DATA}/GEBCO_gridded.nc")
```
